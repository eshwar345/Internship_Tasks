## Q1. Frequency of each word

**Pseudocode:** Split the sentence into words, convert each word to lowercase, and increment a dictionary counter for each word.

**Example call:** `word_frequency("python is easy and python is powerful",)`

**Expected output:** `{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}`

**Additional test:** `word_frequency("red blue red",)`

**Expected additional output:** `{'red': 2, 'blue': 1}`

```python
def word_frequency(sentence):
    counts = {}
    for word in sentence.lower().split():
        counts[word] = counts.get(word, 0) + 1
    return counts

print(word_frequency("python is easy and python is powerful",))
print(word_frequency("red blue red",))
```

## Q2. Frequency of each character

**Pseudocode:** Visit every character and increment its count in a dictionary.

**Example call:** `character_frequency("banana",)`

**Expected output:** `{'b': 1, 'a': 3, 'n': 2}`

**Additional test:** `character_frequency("letter",)`

**Expected additional output:** `{'l': 1, 'e': 2, 't': 2, 'r': 1}`

```python
def character_frequency(word):
    counts = {}
    for character in word:
        counts[character] = counts.get(character, 0) + 1
    return counts

print(character_frequency("banana",))
print(character_frequency("letter",))
```

## Q3. First non-repeating character

**Pseudocode:** Count characters first; then scan the word from left to right and return the first character whose count is one. Return None if none qualifies.

**Example call:** `first_non_repeating_character("swiss",)`

**Expected output:** `'w'`

**Additional test:** `first_non_repeating_character("aabb",)`

**Expected additional output:** `None`

```python
def first_non_repeating_character(word):
    counts = character_frequency(word)
    for character in word:
        if counts[character] == 1:
            return character
    return None

print(first_non_repeating_character("swiss",))
print(first_non_repeating_character("aabb",))
```

## Q4. Check anagrams

**Pseudocode:** Normalize both strings by removing spaces and lowercasing; two strings are anagrams when their character-frequency dictionaries match.

**Example call:** `are_anagrams("listen", "silent")`

**Expected output:** `True`

**Additional test:** `are_anagrams("Dormitory", "Dirty room")`

**Expected additional output:** `True`

```python
def are_anagrams(s1, s2):
    return character_frequency(s1.replace(" ", "").lower()) == character_frequency(
        s2.replace(" ", "").lower()
    )

print(are_anagrams("listen", "silent"))
print(are_anagrams("Dormitory", "Dirty room"))
```

## Q5. First repeating character

**Pseudocode:** Track characters already seen. The first character encountered a second time is returned; if no repeat occurs, return None.

**Example call:** `first_repeating_character("programming",)`

**Expected output:** `'r'`

**Additional test:** `first_repeating_character("abcdef",)`

**Expected additional output:** `None`

```python
def first_repeating_character(text):
    seen = set()
    for character in text:
        if character in seen:
            return character
        seen.add(character)
    return None

print(first_repeating_character("programming",))
print(first_repeating_character("abcdef",))
```

## Q6. Move zeros to the end

**Pseudocode:** Collect non-zero values in their original order, collect zeros, and concatenate the two lists.

**Example call:** `move_zeros_to_end([0, 3, 0, 1, 2, 0, 4],)`

**Expected output:** `[3, 1, 2, 4, 0, 0, 0]`

**Additional test:** `move_zeros_to_end([0, 0, 5, 0],)`

**Expected additional output:** `[5, 0, 0, 0]`

```python
def move_zeros_to_end(nums):
    non_zeros = [number for number in nums if number != 0]
    zeros = [number for number in nums if number == 0]
    return non_zeros + zeros

print(move_zeros_to_end([0, 3, 0, 1, 2, 0, 4],))
print(move_zeros_to_end([0, 0, 5, 0],))
```

## Q7. Second highest distinct number

**Pseudocode:** Remove duplicates, sort values from largest to smallest, and return the second value if it exists; otherwise return None.

**Example call:** `second_highest_distinct([10, 20, 40, 30, 40, 20],)`

**Expected output:** `30`

**Additional test:** `second_highest_distinct([5, 5, 5],)`

**Expected additional output:** `None`

```python
def second_highest_distinct(nums):
    distinct = sorted(set(nums), reverse=True)
    if len(distinct) < 2:
        return None
    return distinct[1]

print(second_highest_distinct([10, 20, 40, 30, 40, 20],))
print(second_highest_distinct([5, 5, 5],))
```

## Q8. Remove duplicates preserving order

**Pseudocode:** Scan the list from left to right and append an item only if it has not been seen before.

**Example call:** `remove_duplicates_preserve_order([4, 2, 4, 5, 2, 7, 5],)`

**Expected output:** `[4, 2, 5, 7]`

**Additional test:** `remove_duplicates_preserve_order([1, 1, 2, 3, 2],)`

**Expected additional output:** `[1, 2, 3]`

```python
def remove_duplicates_preserve_order(nums):
    result = []
    seen = set()
    for number in nums:
        if number not in seen:
            result.append(number)
            seen.add(number)
    return result

print(remove_duplicates_preserve_order([4, 2, 4, 5, 2, 7, 5],))
print(remove_duplicates_preserve_order([1, 1, 2, 3, 2],))
```

## Q9. Missing number from 1 to N

**Pseudocode:** Compute the expected sum 1 + ... + N using N*(N+1)//2; subtract the sum of the supplied list.

**Example call:** `find_missing_number([1, 2, 4, 5, 6], 6)`

**Expected output:** `3`

**Additional test:** `find_missing_number([2, 3, 4, 5], 5)`

**Expected additional output:** `1`

```python
def find_missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

print(find_missing_number([1, 2, 4, 5, 6], 6))
print(find_missing_number([2, 3, 4, 5], 5))
```

## Q10. First pair that sums to target

**Pseudocode:** Scan values from left to right. Before storing the current value, check whether its complement has already appeared; return that pair immediately.

**Example call:** `find_first_pair_sum([2, 7, 11, 15], 9)`

**Expected output:** `[2, 7]`

**Additional test:** `find_first_pair_sum([1, 4, 2, 3], 6)`

**Expected additional output:** `[4, 2]`

```python
def find_first_pair_sum(nums, target):
    seen = set()
    for number in nums:
        needed = target - number
        if needed in seen:
            return [needed, number]
        seen.add(number)
    return []

print(find_first_pair_sum([2, 7, 11, 15], 9))
print(find_first_pair_sum([1, 4, 2, 3], 6))
```

## Q11. Number closest to zero

**Pseudocode:** Choose the value with the smallest absolute value; when absolute values tie, prefer the larger (positive) value.

**Example call:** `closest_to_zero([-8, -3, 2, 5, -2],)`

**Expected output:** `2`

**Additional test:** `closest_to_zero([-4, -1, 1, 7],)`

**Expected additional output:** `1`

```python
def closest_to_zero(nums):
    if not nums:
        return None
    return min(nums, key=lambda number: (abs(number), -number))

print(closest_to_zero([-8, -3, 2, 5, -2],))
print(closest_to_zero([-4, -1, 1, 7],))
```

## Q12. Common elements

**Pseudocode:** Make a set from the second list, then scan the first list and add matching elements once, preserving first-list order.

**Example call:** `common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7])`

**Expected output:** `[3, 4, 5]`

**Additional test:** `common_elements([4, 4, 2, 1], [1, 4])`

**Expected additional output:** `[4, 1]`

```python
def common_elements(list1, list2):
    second_values = set(list2)
    result = []
    seen = set()
    for item in list1:
        if item in second_values and item not in seen:
            result.append(item)
            seen.add(item)
    return result

print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))
print(common_elements([4, 4, 2, 1], [1, 4]))
```

## Q13. Rotate right by K

**Pseudocode:** If the list is non-empty, reduce K modulo the list length. Join the last K elements before the remaining prefix.

**Example call:** `rotate_right([1, 2, 3, 4, 5], 2)`

**Expected output:** `[4, 5, 1, 2, 3]`

**Additional test:** `rotate_right([1, 2, 3], 5)`

**Expected additional output:** `[2, 3, 1]`

```python
def rotate_right(nums, k):
    if not nums:
        return []
    k %= len(nums)
    if k == 0:
        return nums[:]
    return nums[-k:] + nums[:-k]

print(rotate_right([1, 2, 3, 4, 5], 2))
print(rotate_right([1, 2, 3], 5))
```

## Q14. Maximum sum of consecutive pair

**Pseudocode:** Start with the first pair's sum. Check every following adjacent pair and keep the greatest sum.

**Example call:** `max_consecutive_pair_sum([4, 7, 2, 9, 5],)`

**Expected output:** `14`

**Additional test:** `max_consecutive_pair_sum([-5, -2, -8],)`

**Expected additional output:** `-7`

```python
def max_consecutive_pair_sum(nums):
    if len(nums) < 2:
        raise ValueError("At least two elements are required.")
    best = nums[0] + nums[1]
    for index in range(1, len(nums) - 1):
        pair_sum = nums[index] + nums[index + 1]
        if pair_sum > best:
            best = pair_sum
    return best

print(max_consecutive_pair_sum([4, 7, 2, 9, 5],))
print(max_consecutive_pair_sum([-5, -2, -8],))
```

## Q15. All unique pairs summing to target

**Pseudocode:** Check each value's complement, normalize each valid pair into ascending order, store pairs in a set to remove duplicates, then sort the result.

**Example call:** `all_unique_pairs_sum([2, 4, 3, 5, 7, 8, 1], 9)`

**Expected output:** `[[1, 8], [2, 7], [4, 5]]`

**Additional test:** `all_unique_pairs_sum([1, 1, 2, 2, 3, 3], 4)`

**Expected additional output:** `[[1, 3], [2, 2]]`

```python
def all_unique_pairs_sum(nums, target):
    values = set(nums)
    pairs = set()
    for number in nums:
        other = target - number
        if other in values:
            if number != other or nums.count(number) >= 2:
                pairs.add(tuple(sorted((number, other))))
    return [list(pair) for pair in sorted(pairs)]

print(all_unique_pairs_sum([2, 4, 3, 5, 7, 8, 1], 9))
print(all_unique_pairs_sum([1, 1, 2, 2, 3, 3], 4))
```
