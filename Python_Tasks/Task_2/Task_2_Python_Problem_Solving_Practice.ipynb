## Q1. Find the Shortest Word in a Sentence

**1. Observe:** Input is a sentence containing words separated by whitespace.

**2. Understand:** Return the shortest word. If tied, this solution keeps the first shortest word.

**3. Break:** Split the sentence into words, then compare their lengths.

**4. Logic (pseudocode):** Start with the first word as the current shortest. Replace it only when a shorter word is found.

**5. Code:** Implement the steps below using Python.

**6. Test:** Run the given example and the additional test after it.

**Expected example output:** `easy`

**Edge case / note:** An empty sentence has no words; this solution assumes the input contains at least one word.


```python
sentence = "Python makes coding easy"

words = sentence.split()
shortest = words[0]

for word in words:
    if len(word) < len(shortest):
        shortest = word

print(shortest)

# Additional test case
sentence = "We code in Python"
words = sentence.split()
shortest = words[0]
for word in words:
    if len(word) < len(shortest):
        shortest = word
print(shortest)
```

## Q8. Find the Most Frequent Character

**1. Observe:** Spaces are excluded, and ties are resolved by first appearance.

**2. Understand:** Return the non-space character with the highest frequency.

**3. Break:** Build character counts, then scan characters in their original order.

**4. Logic (pseudocode):** Count non-space characters. Choose a new winner only when its count is strictly higher.

**5. Code:** Implement the steps below using Python.

**6. Test:** Run the given example and the additional test after it.

**Expected example output:** `a`

**Edge case / note:** This assumes at least one non-space character exists.


```python
text = "banana"

characters = [character for character in text if character != " "]
counts = {}

for character in characters:
    counts[character] = counts.get(character, 0) + 1

most_frequent = characters[0]
for character in characters:
    if counts[character] > counts[most_frequent]:
        most_frequent = character

print(most_frequent)

# Additional test case
text = "abca"
characters = [character for character in text if character != " "]
counts = {}
for character in characters:
    counts[character] = counts.get(character, 0) + 1
most_frequent = characters[0]
for character in characters:
    if counts[character] > counts[most_frequent]:
        most_frequent = character
print(most_frequent)
```

## Q20. Find the Longest Word That Appears Only Once

**1. Observe:** Words are compared case-insensitively; among unique words, ties go to the first longest word.

**2. Understand:** Find the longest word whose normalized form occurs exactly once.

**3. Break:** Normalize and count words, then inspect only words with frequency one.

**4. Logic (pseudocode):** Keep the first unique word as candidate; replace it only with a strictly longer unique word.

**5. Code:** Implement the steps below using Python.

**6. Test:** Run the given example and the additional test after it.

**Expected example output:** `coding`

**Edge case / note:** If no word occurs exactly once, the program prints None.


```python
sentence = "Python makes coding useful Python"

words = [word.lower() for word in sentence.split()]
counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

longest_unique = None
for word in words:
    if counts[word] == 1:
        if longest_unique is None or len(word) > len(longest_unique):
            longest_unique = word

print(longest_unique)

# Additional test case
sentence = "Alpha beta GAMMA delta"
words = [word.lower() for word in sentence.split()]
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1
longest_unique = None
for word in words:
    if counts[word] == 1:
        if longest_unique is None or len(word) > len(longest_unique):
            longest_unique = word
print(longest_unique)
```
